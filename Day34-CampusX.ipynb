Working with Dates

```python
#Converting to datetime datatype
orders['date'] = pd.to_datetime(orders['date'])
```

```python
orders.info()
```

1. Extract year

```python
orders['date_year'] = orders['date'].dt.year

orders.sample(5)
```

2. Extract Month

```python
orders['date_month_no'] = orders['date'].dt.month

orders.head()
```

```python
orders['date_month_name'] = orders['date'].dt.month_name()

orders.head()
```

Extract Days

```python
orders['date_day'] = orders['date'].dt.day

orders.head()
```

```python
#Day of week
orders['date_dow'] = orders['date'].dt.dayofweek
orders.head()
```

```python
#day of week - name

orders['date_dow_name'] = orders['date'].dt.day_name()
orders.head()
```

```python
# is weekend?

orders['date_is_weekend'] = np.where(orders['date_dow_name'].isin(['Sunday','Saturday']), 1,0)

orders.head()
```

Extract week of the year

```python
orders['date_week'] = orders['date'].dt.isocalendar().week

orders.head()
```

Extract Quarter

```python
orders['quarter'] = orders['date'].dt.quarter

orders.head()
```

Extract Semester

```python
orders['semester'] = np.where(orders['quarter'].isin([1,2]), 1, 2)

orders.head()
```

Extract Time elapsed between dates

```python
import datetime

today = datetime.datetime.today()

today
```

```python
today - orders['date'] 
```

```python
(today - orders['date']).dt.days
```

```python
# Months passed

np.round((today - orders['date']) / np.timedelta64(1, 'W'),0) #need to do it for months
```
