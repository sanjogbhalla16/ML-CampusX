```python
import pandas as pd
import seaborn as sns
```

```python
tips = sns.load_dataset('tips')
tips
```

```python
titanic = pd.read_csv('Titanic-Dataset.csv')
titanic
```

```python
flights = sns.load_dataset('flights')
```

```python
iris = sns.load_dataset('iris')
```

```python
#Scatterplot (Numerical - Numerical)
sns.scatterplot(data = tips, x = 'total_bill', y = 'tip', hue='sex', style='smoker',size='size') #linear relationship
```

```python
#Bar Plot - (Numerical - Categorical)
#according to Pclass, what is the average age of passengers?
sns.barplot(data = titanic, x = 'Pclass', y = 'Fare', hue='Sex') #that line is confidence interval
```

```python
#Box Plot - (Numerical - Categorical)
sns.boxplot(data = titanic, x = 'Sex', y = 'Age', hue='Survived')
```

```python
#Distplot - (Numerical - Categorical)
sns.displot(data = titanic, x = 'Age', hue='Survived' , kind='kde', fill=False)
```

```python
#Heatmap - (Categorical - Categorical)
sns.heatmap(pd.crosstab(titanic['Pclass'], titanic['Survived']))
```

```python
#lets talk in percentage values
titanic.groupby('Pclass').mean(numeric_only=True)['Survived']
```
