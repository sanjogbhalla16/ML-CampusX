```python
from sklearn.datasets import make_regression
import numpy as np
```

```python
X,y = make_regression(n_samples=4, n_features=1, n_informative=1, noise=80, random_state=13)
```

```python
X
```

```python
y
```

```python
import matplotlib.pyplot as plt
plt.scatter(X, y)
```

```python
# Lets apply OLS - Ordinary Least Squares
from sklearn.linear_model import LinearRegression
```

```python
reg = LinearRegression()
```

```python
reg.fit(X,y)
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
plt.scatter(X, y)
plt.plot(X, reg.predict(X), color='red')
```

```python
#lets apply Gradient Descent assuming slope is constant m = 78.35
# and let's assume the starting value of the intercept b = 0
y_pred = ((78.35 * X) + 0).reshape(4)
```

```python
from cProfile import label


plt.scatter(X, y)
plt.plot(X, reg.predict(X), color='red',label='OLS Prediction')
plt.plot(X,y_pred, color="#008cb6", label='Gradient Descent Prediction (b = 0)')
plt.legend()
plt.show()
```

```python
# We are taking out the slope
m = 78.35
b = 0

loss_slope = -2 * np.sum(y - m*X.ravel() - b) #used to flatten a multi-dimensional array into a one-dimensional array. 
loss_slope
```

```python
# Lets take the learning rate = 0.1
lr = 0.1

step_size = loss_slope * lr
step_size
```

```python
# Calculate the new intercept
b = b - step_size
b
```

```python
y_pred1 = ((78.35 * X) + b).reshape(4)

plt.scatter(X, y)
plt.plot(X, reg.predict(X), color='red',label='OLS Prediction')
plt.plot(X,y_pred1, color="#00b61e", label='Gradient Descent Prediction (b = {})'.format((b)))
plt.plot(X,y_pred, color="#15D3DA", label='Gradient Descent Prediction (b = 0)')
plt.legend()
plt.show()
```

```python
# Iteration 2
loss_slope = -2 * np.sum(y - m*X.ravel() - b)
step_size
```

```python
step_size = loss_slope * lr
step_size
```

```python
b = b - step_size
b
```

```python
y_pred2 = ((78.35 * X) + b).reshape(4)

plt.scatter(X, y)
plt.plot(X, reg.predict(X), color='red',label='OLS Prediction')
plt.plot(X,y_pred2, color="#064047", label='Gradient Descent Prediction (b = {})'.format((b)))
plt.plot(X,y_pred1, color="#00b61e", label='Gradient Descent Prediction (b = {})'.format((b)))
plt.plot(X,y_pred, color="#15D3DA", label='Gradient Descent Prediction (b = 0)')
plt.legend()
plt.show()
```

```python
# Iteration 3
loss_slope = -2 * np.sum(y - m*X.ravel() - b)
loss_slope
```

```python
step_size = loss_slope * lr
step_size
```

```python
b = b - step_size
b
```

```python
y_pred3 = ((78.35 * X) + b).reshape(4)

plt.scatter(X, y)
plt.plot(X, reg.predict(X), color='red',label='OLS Prediction')
plt.plot(X,y_pred3, color="#191CE6", label='Gradient Descent Prediction (b = {})'.format((b)))
plt.plot(X,y_pred2, color="#50C6D3", label='Gradient Descent Prediction (b = {})'.format((b)))
plt.plot(X,y_pred1, color="#00b61e", label='Gradient Descent Prediction (b = {})'.format((b)))
plt.plot(X,y_pred, color="#15D3DA", label='Gradient Descent Prediction (b = 0)')
plt.legend()
plt.show()
```

```python
b = -100
m = 78.35
lr = 0.01

epochs = 100

for i in range(epochs):
  loss_slope = -2 * np.sum(y - m*X.ravel() - b)
  b = b - (lr * loss_slope)

  y_pred = m * X + b

  plt.plot(X,y_pred)

plt.scatter(X,y)
```

```python
b = -100
m = 78.35
lr = 0.01

epochs = 10

for i in range(epochs):
  loss_slope = -2 * np.sum(y - m*X.ravel() - b)
  b = b - (lr * loss_slope)

  y_pred = m * X + b

  plt.plot(X,y_pred)

plt.scatter(X,y)
```
