```python
company = soup.find_all('div',class_='companyCardWrapper')
```

```python
len(company)
```

```python
names = []
ratings = []
reviews = []

for i in company:
    names.append(i.find('h2').text.strip())
    ratings.append(i.find('div',class_='rating_star_container').text.strip())
    reviews.append(i.find('span',class_='companyCardWrapper__ActionCount').text.strip())
    
d = {"name":names,"rating":ratings,"reviews":reviews}

df = pd.DataFrame(d)
```

```python
# we will create the final dataframe
final = pd.DataFrame()

for j in range(1,11):
    url = 'https://www.ambitionbox.com/list-of-companies?campaign=desktop_nav&page={}'.format(j)
    headers={'User-Agent':'Mozilla/5.0 (Windows NT 6.3; Win 64 ; x64) Apple WeKit /537.36(KHTML , like Gecko) Chrome/80.0.3987.162 Safari/537.36'}
    webpage = requests.get(url,headers = headers).text
    
    soup = BeautifulSoup(webpage,'lxml')
    
    company = soup.find_all('div',class_='companyCardWrapper')
    
    names = []
    ratings = []
    reviews = []

    for i in company:
        names.append(i.find('h2').text.strip())
        ratings.append(i.find('div',class_='rating_star_container').text.strip())
        reviews.append(i.find('span',class_='companyCardWrapper__ActionCount').text.strip())
        
    d = {"name":names,"rating":ratings,"reviews":reviews}

    final = pd.concat([final, df], ignore_index=True)   # ✅ append to final DataFrame
```
